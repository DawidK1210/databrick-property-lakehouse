```python
%sql
CREATE OR REPLACE TABLE workspace.property.gold_suburb_summary AS
SELECT
  suburb,
  city,
  COUNT(*) AS listings,
  COUNT(asking_price) AS listings_with_price,
  ROUND(MEDIAN(asking_price)) AS median_price,
  ROUND(AVG(asking_price / floor_size_sqm)) AS avg_price_per_sqm
FROM workspace.property.silver_listings
WHERE asking_price IS NOT NULL AND floor_size_sqm IS NOT NULL
GROUP BY suburb, city;
```

```python
%sql
CREATE OR REPLACE TABLE workspace.property.gold_price_by_type AS
SELECT
  suburb,
  property_type,
  COUNT(*) AS listings,
  ROUND(MEDIAN(asking_price)) AS median_price
FROM workspace.property.silver_listings
WHERE asking_price IS NOT NULL
GROUP BY suburb, property_type;
```

```python
%sql
CREATE OR REPLACE TABLE workspace.property.gold_monthly_listings AS
SELECT
  date_trunc('month', listing_date) AS month,
  COUNT(*) AS new_listings,
  ROUND(MEDIAN(asking_price)) AS median_price
FROM workspace.property.silver_listings
GROUP BY date_trunc('month', listing_date)
ORDER BY month;
```

```python
%sql
SELECT * FROM workspace.property.gold_suburb_summary
ORDER BY median_price DESC;
```
