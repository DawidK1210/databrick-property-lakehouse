```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.property;
CREATE VOLUME IF NOT EXISTS workspace.property.raw;
```

```python
df = (spark.read
      .option("header", True)
      .option("inferSchema", True)
      .csv("/Volumes/workspace/property/raw/property_listings.csv"))

display(df.limit(10))

df.write.mode("overwrite").saveAsTable("workspace.property.bronze_listings")
```

```python
%sql
SELECT COUNT(*) AS total_rows,
       COUNT(DISTINCT listing_id) AS unique_listings,
       COUNT(DISTINCT suburb) AS distinct_suburbs
FROM workspace.property.bronze_listings;
```
